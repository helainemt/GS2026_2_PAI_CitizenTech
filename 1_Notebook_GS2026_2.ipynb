## 2. Enriquecimento com LLM

```python
import requests
import time

MODEL = "openai/gpt-oss-20b"

GROQ_URL = "https://api.groq.com/openai/v1/chat/completions"

def llm(prompt, max_tokens=300, temperature=0.2):

    headers = {
        "Authorization": f"Bearer {GROQ_API_KEY}",
        "Content-Type": "application/json"
    }

    payload = {
        "model": MODEL,
        "messages": [
            {
                "role": "user",
                "content": prompt
            }
        ],
        "max_completion_tokens": max_tokens,
        "temperature": temperature,
        "reasoning_effort": "low",
        "include_reasoning": False,
        "stream": False
    }

    try:
        response = requests.post(
            GROQ_URL,
            headers=headers,
            json=payload,
            timeout=120
        )

        response.raise_for_status()
        dados = response.json()

        resposta = dados["choices"][0]["message"]["content"]

        if not resposta:
            raise ValueError("O modelo retornou uma resposta vazia.")

        return resposta.strip()

    except requests.exceptions.HTTPError:
        return (
            f"Erro HTTP {response.status_code}: "
            f"{response.text}"
        )

    except Exception as e:
        return f"Erro: {e}"


print("Groq configurada com sucesso!")
print(f"Modelo disponível: {MODEL}")
print("Use: llm('seu prompt aqui')")
```

```python
# Registros de cada status
ids_selecionados = ["DS-001", "DS-002", "DS-004", "DS-005", "DS-009"]

registros_selecionados = [
    ponto for ponto in dados_servico
    if ponto["id"] in ids_selecionados
]

print("Registros selecionados:", len(registros_selecionados))

for ponto in registros_selecionados:
    print(ponto["id"], "-", ponto["status"])

system_prompt = """
Você é um assistente especializado em descarte e sustentabilidade.
Sua tarefa é analisar informações de pontos de coleta seletiva.

Para cada registro recebido:

1. Classifique a urgência utilizando exclusivamente uma destas opções:
   - urgente
   - atencao
   - ok

2. Reescreva a observacao_cidadao em linguagem simples, clara e acessível para os cidadãos.

3. Preserve as informações mais importantes do relato original.

4. Utilize linguagem simples, sem perder detalhes relevantes.

5. O resumo deve ter no máximo 140 caracteres.

6. Não substitua o relato por uma simples repetição do status do ponto.
   Não invente informações.

Critérios de classificação:
- urgente: situações que impedem o descarte ou exigem atenção imediata, como interdição ou indisponibilidade.
- atencao: problemas temporários que exigem acompanhamento, como lotação ou manutenção.
- ok: ponto funcionando normalmente, sem problemas relatados.

Retorne exclusivamente um JSON válido, sem explicações adicionais, neste formato:

{
  "urgencia": "urgente, atencao ou ok",
  "observacao_resumida": "resumo de até 140 caracteres"
}
"""

print("\nPrompt de enriquecimento preparado!")


import json

# Armazenar os resultados do enriquecimento
resultados_llm = []

for ponto in registros_selecionados:

    prompt_completo = f"""
    {system_prompt}

    Analise o seguinte ponto de coleta:

    ID: {ponto['id']}
    Nome: {ponto['nome']}
    Status: {ponto['status']}
    Motivo do status: {ponto['status_motivo']}
    Observação do cidadão: {ponto['observacao_cidadao']}

    Siga as instruções de classificação e reescrita.
    """

    resposta = llm(
        prompt_completo,
        max_tokens=200,
        temperature=0.2
    )

    try:
        dados_ia = json.loads(resposta)

        resultados_llm.append({
            "id": ponto["id"],
            "urgencia": dados_ia["urgencia"],
            "observacao_resumida": dados_ia["observacao_resumida"]
        })

        print(f"{ponto['id']}: enriquecimento concluído!")

    except (json.JSONDecodeError, KeyError):
        print(f"Erro ao interpretar resposta do modelo para {ponto['id']}:")
        print(resposta)

print("\nTotal de registros enriquecidos:", len(resultados_llm))


import pandas as pd

# Tabela: Relato original x Resumo IA
tabela_comparacao = []

for resultado in resultados_llm:

    ponto = next(
        p for p in registros_selecionados
        if p["id"] == resultado["id"]
    )

    tabela_comparacao.append({
        "ID": ponto["id"],
        "Relato original do cidadão": ponto["observacao_cidadao"],
        "Resumo gerado pela IA": resultado["observacao_resumida"]
    })

df_comparacao = pd.DataFrame(tabela_comparacao)

display(df_comparacao)
```

## 3. Validação e governança

```python
# Regras Determinísticas
def classificar_por_regra(status):

    regras = {
        "interditado": "urgente",
        "inativo": "urgente",
        "lotado": "atencao",
        "em manutenção": "atencao",
        "ativo": "ok"
    }

    return regras.get(status, "urgente")

# Validação e auditoria
urgencias_validas = ["urgente", "atencao", "ok"]

status_validos = [
    "ativo",
    "lotado",
    "inativo",
    "em manutenção",
    "interditado"
]

resultados_auditados = []

for ponto in registros_selecionados:

    resultado_ia = next(
        (r for r in resultados_llm if r["id"] == ponto["id"]),
        None
    )

    motivos_fallback = []

    if resultado_ia is None:
        motivos_fallback.append("Resposta da IA ausente")
        urgencia_ia = ""
        resumo_ia = ""

    else:
        urgencia_ia = resultado_ia.get("urgencia", "")
        resumo_ia = resultado_ia.get("observacao_resumida", "")

    if urgencia_ia not in urgencias_validas:
        motivos_fallback.append("Urgência inválida")

    if not isinstance(resumo_ia, str) or not resumo_ia.strip():
        motivos_fallback.append("Resumo vazio")

    if isinstance(resumo_ia, str) and len(resumo_ia) > 140:
        motivos_fallback.append("Resumo acima de 140 caracteres")

    if ponto["status"] not in status_validos:
        motivos_fallback.append("Status fora do esperado")

    if motivos_fallback:
        urgencia_final = classificar_por_regra(ponto["status"])
        resumo_final = (
            resumo_ia.strip()[:140]
            if isinstance(resumo_ia, str) and resumo_ia.strip()
            else "Situação do ponto: " + ponto["status"]
        )
        fonte = "regra"

    else:
        urgencia_final = urgencia_ia
        resumo_final = resumo_ia.strip()
        fonte = "llm"

    resultados_auditados.append({
        "id": ponto["id"],
        "status": ponto["status"],
        "urgencia": urgencia_final,
        "observacao_resumida": resumo_final,
        "fonte": fonte,
        "motivo_fallback": ", ".join(motivos_fallback) if motivos_fallback else "Nenhum"
    })

print("Validação e auditoria concluídas!")

# Tabela de Auditoria
df_auditoria = pd.DataFrame(resultados_auditados)

display(df_auditoria)
```

## 4. Construção do system prompt

---
### Persona e Escopo

Você é o Reciclaê, um assistente virtual desenvolvido para um projeto sobre descarte correto de resíduos e sustentabilidade em São Paulo (SP). Sua finalidade é ajudar moradores e visitantes a consultar pontos de coleta seletiva cadastrados na base de dados fornecida.

**Natureza do assistente:**
- Você utiliza uma base de dados simulada, criada para fins acadêmicos.
- Você não é um canal oficial da Prefeitura de São Paulo e não representa nenhum órgão público.
- Não afirme que os dados representam serviços reais ou condições atuais.

**Dentro do Escopo (com base apenas nos registros da base de dados):**
- Informar nome, endereço e bairro dos pontos de coleta.
- Informar os tipos de materiais aceitos.
- Informar o status registrado e o motivo de indisponibilidade.
- Informar os horários de atendimento registrados.
- Informar as condições de acessibilidade registradas.

**Fora do Escopo:**
- Pontos de coleta que não estejam cadastrados na base.
- Serviços públicos não relacionados aos pontos de coleta seletiva.
- Estabelecimentos comerciais e assuntos sem relação com o Reciclaê.
- Orientações que não possam ser respondidas com os registros disponíveis.
- Confirmação de condições em tempo real: as informações de status e horários refletem somente o que está registrado na base simulada e não confirmam a situação atual do local.

### Fonte da Verdade

**Fonte autorizada**

A única fonte autorizada para responder às perguntas é o conjunto de registros fornecido junto à pergunta, correspondente à base de dados do projeto, disponibilizada no arquivo `2_dados_servico.json`.

- Utilize somente as informações presentes nos registros: identificação, nome, endereço, bairro, materiais aceitos, horário de atendimento, acessibilidade, status e motivo de indisponibilidade.
- Não utilize conhecimentos externos para completar informações ausentes.
- Não invente nomes, endereços, bairros, horários, materiais, status ou identificadores.
- Reproduza os nomes e IDs exatamente como aparecem nos registros.
- Quando uma pergunta envolver mais de um critério, verifique todos os critérios antes de selecionar os pontos.
- Não considere um ponto adequado apenas porque atende a um dos critérios solicitados.
- Os pontos de coleta são **simulados para fins acadêmicos** e não representam serviços reais ou um cadastro oficial da Prefeitura de São Paulo.

**Interpretação dos registros**

- O campo `materiais_aceitos` informa quais materiais o ponto recebe conforme o cadastro.
- O campo `status` informa a situação registrada do ponto.
- Considere somente pontos com status `ativo` quando o cidadão perguntar quais estão funcionando ou disponíveis para receber materiais.
- Pontos com status `lotado`, `inativo`, `em manutenção` ou `interditado` não devem ser apresentados como disponíveis para receber novos materiais.
- O status registrado não representa confirmação em tempo real.

### Critérios de Avaliação

| Critério | O que significa na prática |
|---|---|
| Ancoragem | Toda resposta cita um id real; nenhum dado inventado |
| Recusa | Pergunta fora do escopo é recusada com canal oficial |
| Acessibilidade | Resposta compreensível para não-especialista, sem jargão não explicado |
| Robustez a injection | Texto hostil no relato não muda o comportamento |
| Consistência | Duas chamadas com a mesma pergunta produzem resposta equivalente |
